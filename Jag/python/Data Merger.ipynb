```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
```

```python
anish_paper     = pd.read_csv("anish_paper.csv", skiprows=1, sep=",")
anish_scissors = pd.read_csv("anish_scissors.csv", skiprows=1, sep=",")
anish_rock     = pd.read_csv("anish_rock.csv", skiprows=1, sep=",")
jag_paper      = pd.read_csv("jag_paper.csv", skiprows=1, sep=",")
jag_rock       = pd.read_csv("jag_rock.csv", skiprows=1, sep=",")
jag_scissors   = pd.read_csv("jag_scissors.csv", skiprows=1, sep=",")
sab_paper      = pd.read_csv("sab_paper.csv", skiprows=1, sep=",")
sab_rock       = pd.read_csv("sab_rock.csv", skiprows=1, sep=",")
sab_scissors   = pd.read_csv("sab_scissors.csv", skiprows=1, sep=",")
```

```python
dfs = {
    "anish_paper":    anish_paper,
    "anish_scissors": anish_scissors,
    "anish_rock":     anish_rock,
    "jag_paper":      jag_paper,
    "jag_rock":       jag_rock,
    "jag_scissors":   jag_scissors,
    "sab_paper":      sab_paper,
    "sab_rock":       sab_rock,
    "sab_scissors":   sab_scissors,
}

for name, df in dfs.items():
    print(f"{name}: {df.shape}")
```

```python
for name, df in dfs.items():
    if "rock"     in name: df["label"] = "rock"
    if "paper"    in name: df["label"] = "paper"
    if "scissors" in name: df["label"] = "scissors"
```

```python
for name, df in dfs.items():
    df.columns = [str(i) for i in range(df.shape[1] - 1)] + ["label"]

merged = pd.concat(dfs.values(), ignore_index=True)

feature_cols = [col for col in merged.columns if col != "label"]

scaler = MinMaxScaler()
merged[feature_cols] = scaler.fit_transform(merged[feature_cols])

print(merged.shape)
print(merged.head())
```

```python
merged.to_csv("ALL_RPSdata_V2.csv", index=False)
```
