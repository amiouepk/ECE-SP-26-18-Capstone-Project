```python
# Imports
import re
import math
import numpy as np
import pandas as pd
from pathlib import Path
```

```python
def normalize_sensor_data(match):
    # Extract the dynamic labels (e.g., 'X', 'gyro_x', 'mag_x')
    label_x = match.group(1)
    label_y = match.group(3)
    label_z = match.group(5)

    # Extract the raw numeric values from the regex match
    x = float(match.group(2))
    y = float(match.group(4))
    z = float(match.group(6))
    
    # Calculate the vector magnitude
    magnitude = math.sqrt(x**2 + y**2 + z**2)
    
    # Safety check to avoid division by zero
    if magnitude == 0:
        return match.group(0) # Returns the exact original string
        
    # Calculate the unit vector components
    nx = x / magnitude
    ny = y / magnitude
    nz = z / magnitude
    
    # Return the formatted string using the original labels, rounded to 3 decimal places 
    return f"{label_x}: {nx:.3f} | {label_y}: {ny:.3f} | {label_z}: {nz:.3f}"

def normalize_sensor_log(data_text):
    # Matches any text label (X, gyro_x, mag_x), followed by a colon and the number.
    # Groups 1, 3, 5 capture the labels. Groups 2, 4, 6 capture the numeric values.
    pattern = r"([a-zA-Z_]+):\s*([-0-9.]+)\s*\|\s*([a-zA-Z_]+):\s*([-0-9.]+)\s*\|\s*([a-zA-Z_]+):\s*([-0-9.]+)"
    
    # Substitute matches using the updated normalization function
    normalized_data = re.sub(pattern, normalize_sensor_data, data_text)
    return normalized_data
```

```python
def parse_text_to_dataframe(raw_text, label_name):
    """Parses raw sensor text into a flattened pandas DataFrame."""
    # Split into time blocks
    blocks = re.split(r'(?=Time:\s)', raw_text.strip())
    blocks = [b.strip() for b in blocks if b.strip()]

    # We will build a list of dictionaries, where each dict is one row in our CSV
    rows = []

    for block in blocks:
        lines = block.strip().split('\n')
        time_match = re.match(r'Time:\s*([\d.]+)', lines[0])
        if not time_match:
            continue
        
        # Initialize the row with the timestamp and the label
        row_data = {
            'time': float(time_match.group(1)),
            'label': label_name
        }

        # Find each sensor's data and add it to the row
        for sensor_id in range(1, 7):
            pattern = rf'Sensor {sensor_id}:\s*\nX:\s*([-\d.]+)\s*\|\s*Y:\s*([-\d.]+)\s*\|\s*Z:\s*([-\d.]+)\s*\ngyro_x:\s*([-\d.]+)\s*\|\s*gyro_y:\s*([-\d.]+)\s*\|\s*gyro_z:\s*([-\d.]+)(?:\s*\nmag_x:\s*([-\d.]+)\s*\|\s*mag_y:\s*([-\d.]+)\s*\|\s*mag_z:\s*([-\d.]+))?'
            m = re.search(pattern, block)
            
            if m:
                row_data[f's{sensor_id}_X'] = float(m.group(1))
                row_data[f's{sensor_id}_Y'] = float(m.group(2))
                row_data[f's{sensor_id}_Z'] = float(m.group(3))
                row_data[f's{sensor_id}_gyro_x'] = float(m.group(4))
                row_data[f's{sensor_id}_gyro_y'] = float(m.group(5))
                row_data[f's{sensor_id}_gyro_z'] = float(m.group(6))
                
                # Handle optional magnetometer data
                if m.group(7):
                    row_data[f's{sensor_id}_mag_x'] = float(m.group(7))
                    row_data[f's{sensor_id}_mag_y'] = float(m.group(8))
                    row_data[f's{sensor_id}_mag_z'] = float(m.group(9))
                else:
                    row_data[f's{sensor_id}_mag_x'] = np.nan
                    row_data[f's{sensor_id}_mag_y'] = np.nan
                    row_data[f's{sensor_id}_mag_z'] = np.nan

        rows.append(row_data)

    # Convert the list of rows into a pandas DataFrame
    df = pd.DataFrame(rows)
    
    # Reorder columns so 'label' is at the very end (standard ML practice)
    cols = [c for c in df.columns if c != 'label'] + ['label']
    df = df[cols]
    
    return df
```

```python
file = "anish-scissors-1_50hz"
label = "scissors"
```

```python
# Create a robust path to your file (e.g., if it's in a 'data' subfolder)
file_path = Path(f'data/{file}.txt')

# Check if the file actually exists before trying to read it
if file_path.exists():
    # .read_text() opens, reads, and closes the file all in one step
    raw_data = file_path.read_text()
    print(f"Success: Loaded {len(raw_data)} characters.")
else:
    print(f"Error: Could not find file at {file_path.absolute()}")
```

```python
#Running the function
normalized_text = normalize_sensor_log(raw_data)
```

```python
normalized_file_path = Path(f'data/{file}_normalized.txt')
normalized_file_path.write_text(normalized_text)
print(f"Normalized data saved to {normalized_file_path.absolute()}")
```

```python
# Create a robust path to your file (e.g., if it's in a 'data' subfolder)
file_path = Path(f'data/{file}_normalized.txt')

# Check if the file actually exists before trying to read it
if file_path.exists():
    # .read_text() opens, reads, and closes the file all in one step
    raw_data = file_path.read_text()
    print(f"Success: Loaded {len(raw_data)} characters.")
else:
    print(f"Error: Could not find file at {file_path.absolute()}")
```

```python

# 1. Parse the text into a DataFrame, using the filename as the label
df = parse_text_to_dataframe(raw_data, label_name=label)
    
# 2. Save it to a CSV file
csv_path = Path(f'data/{file}_normalized_dataset.csv')
df = df.dropna(axis=1, how='all')
df.to_csv(csv_path, index=False) 
print(f"Success: Saved CSV to {csv_path}")

# 3. Display the first 5 rows in the notebook
display(df.head())
```

# After this is for combining different CSVs and then making one big file and removing empty columns

```python
# 1. Define the list of CSV files you want to combine
# Make sure these match the names of the files you generated earlier
csv_files = ['data/jag-random-no-poses-50hz_normalized_dataset.csv', 'data/jag-scissors-1_50hz_normalized_dataset.csv', 'data/jag-paper-1_50hz_normalized_dataset.csv', 'data/jag-rock-1_50hz_normalized_dataset.csv']
```

```python
# 2. Create an empty list to hold the dataframes
dataframes = []
```

```python
# 3. Loop through the list, load each CSV, and append it to our list
for file in csv_files:
    file_path = Path(file)
    if file_path.exists():
        df = pd.read_csv(file_path)
        dataframes.append(df)
        print(f"Loaded {file} with {len(df)} rows.")
    else:
        print(f"Warning: Could not find {file}")
```

```python
# 4. Combine all the dataframes into one master dataframe
if dataframes:
    # ignore_index=True resets the row numbers so they go continuously from 0 to the end
    master_df = pd.concat(dataframes, ignore_index=True)
    print(f"\nCombined Dataset Shape (Rows, Columns): {master_df.shape}")
    
    # 5. Clean empty columns
    # dropna(axis=1, how='all') drops any column where ALL the values are NaN (empty)
    master_df = master_df.dropna(axis=1, how='all')
    print(f"Cleaned Dataset Shape (Rows, Columns): {master_df.shape}")
    
    # --- NEW SHUFFLING STEP ---
    # sample(frac=1) pulls 100% of the rows in a random order.
    # random_state=42 ensures the shuffle is exactly the same every time you run this cell.
    # reset_index(drop=True) gives the shuffled rows clean, ordered index numbers (0, 1, 2...).
    master_df = master_df.sample(frac=1, random_state=42).reset_index(drop=True)
    print("Successfully shuffled the dataset rows.")
    # --------------------------
    
    # 6. Save the final combined dataset
    output_path = 'data/combined_training_data.csv'
    master_df.to_csv(output_path, index=False)
    print(f"Success! Saved combined data to {output_path}")
    
    # Preview the final data
    display(master_df.head())
    display(master_df.tail())
else:
    print("No dataframes were loaded. Check your file paths.")
```
